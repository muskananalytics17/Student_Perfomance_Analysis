# Train Model

Sperate features and target column

```python
x = df.drop(columns=['math_score'])
y = df[['math_score']]
```

Encoding

```python
from sklearn.preprocessing import LabelEncoder
```

```python
for i in x.columns:
    if x[i].dtype=='object':
        x = pd.get_dummies(x, columns=[i],drop_first=False,dtype=int)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
from sklearn.preprocessing import StandardScaler
sc= StandardScaler()
x_train_scaled = sc.fit_transform(x)
x_test_scaled = sc.transform(x)
```

```python
y_test=pd.DataFrame(y_test)
```

```python
y_test=[y_test]
```

```python
from sklearn.linear_model import LinearRegression,Lasso,Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
def evaluate_model(true,pred):
    mse=mean_squared_error(true,pred)
    mae=mean_absolute_error(true,pred)
    rmse=np.sqrt(mse)
    r2=r2_score(true,pred)

    return mse,mae,rmse,r2

      
```

```python
print(x_train_scaled.shape)
print(y_train.shape)
```

```python
models= {
    'Linear Regression': LinearRegression(),
    'Redge Regression': Ridge(),
    'Lasso Regression': Lasso(),
    'KNN Regression': KNeighborsRegressor(),
    'Decission Tree Regression': DecisionTreeRegressor(),
    'Random Forest Regression': RandomForestRegressor(),
    'XGB Regression': XGBRegressor(),
    'AdaBoost Regression':AdaBoostRegressor()
}
model_ist=[]
r2_list=[]
for i in range(len(list(models))):
    model =list(models.values())[i]
    model.fit(x_train,y_train)

    #make predictions
    y_train_pred = model.predict(x_train)
    y_test_pred =model.predict(x_test)

    #evaluation of the model

    mse,mae,rmse,r2=evaluate_model(y_test,y_test_pred)
    mse1,mae1,rmse1,r21= evaluate_model(y_train,y_train_pred)

    r2_list.append(r2)
    model_ist.append(list(models.keys())[i])







```
